```python
import numpy as np

arr_ones = np.ones((2, 3))
arr_full = np.full((2, 2), 7)
arr_even = np.arange(2, 11, 2)

print("np.ones((2, 3)):\n", arr_ones)
print("\nnp.full((2, 2), 7):\n", arr_full)
print("\nnp.arange(2, 11, 2):", arr_even)

zeros_arr = np.zeros((2, 3, 4))
print("ndim:", zeros_arr.ndim)
print("shape:", zeros_arr.shape)
print("size:", zeros_arr.size)

sample_arr = np.array([1, 2.5, 3])
print("dtype:", sample_arr.dtype)
```
