```python
import numpy as np

scores = np.array([
    [85, 90, 78, 92],
    [88, 76, 95, 89],
    [90, 82, 84, 91]
])

print(scores / 5)

print("Student averages (axis=1):", scores.mean(axis=1))
print("Subject max marks (axis=0):", scores.max(axis=0))

class_avg = scores.mean()
student_avgs = scores.mean(axis=1)
mask = student_avgs > class_avg
print(f"Class Average: {class_avg:.2f}")
print("Filtered students/rows:\n", scores[mask])

passing_count = (scores >= 60).sum()
print("Number of scores >= 60:", passing_count)
```

```python
import numpy as np

scores = np.array([
    [85, 90, 78, 92],
    [88, 76, 95, 89],
    [90, 82, 84, 91]
])

pairs = [
    ((4, 3), (3,)),
    ((4, 3), (4,)),
    ((4, 1), (1, 5))
]

for s1, s2 in pairs:
    try:
        res = np.ones(s1) + np.ones(s2)
        print(f"Pair {s1} + {s2} works! Result shape: {res.shape}")
    except ValueError as e:
        print(f"Pair {s1} + {s2} failed: {e}")

bonus = np.array([0, 5, 10]).reshape(3, 1)
print("Scores with individual student bonuses:\n", scores + bonus)

student_means = scores.mean(axis=1, keepdims=True)
centered_scores = scores - student_means
print("Centered scores per student:\n", np.round(centered_scores, 1))
```

```python
import numpy as np

scores = np.array([
    [85, 90, 78, 92],
    [88, 76, 95, 89],
    [90, 82, 84, 91]
])
print(scores[-1, 0:2])
print(scores[0:2, 2])
print(scores[0:2, -2:])
view = scores[0, :2]
view[0] = 999
print("Original modified by view:", scores[0])
scores[0, 0] = 85
safe_copy = scores[0, :2].copy()
safe_copy[0] = 999
print("Original unaffected by copy:", scores[0])
```
