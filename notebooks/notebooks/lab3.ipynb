```python
import numpy as np

scores = np.array([
    [85, 90, 78, 92],
    [88, 76, 95, 89],
    [90, 82, 84, 91]
])
print(scores[-1, 0:2])
print(scores[0:2, 2])
print(scores[0:2, -2:])
view = scores[0, :2]
view[0] = 999
print("Original modified by view:", scores[0])
scores[0, 0] = 85
safe_copy = scores[0, :2].copy()
safe_copy[0] = 999
print("Original unaffected by copy:", scores[0])
```

```python
my_view = scores[0, :2]
my_view[0] = 999
print("Original after view change:", scores[0])

scores[0, 0] = 85 

my_copy = scores[0, :2].copy()
my_copy[0] = 999
print("Original after copy change:", scores[0]) 
```

```python
import numpy as np

arr_ones = np.ones((2, 3))
arr_full = np.full((2, 2), 7)
arr_even = np.arange(2, 11, 2)

print("np.ones((2, 3)):\n", arr_ones)
print("\nnp.full((2, 2), 7):\n", arr_full)
print("\nnp.arange(2, 11, 2):", arr_even)

zeros_arr = np.zeros((2, 3, 4))
print("ndim:", zeros_arr.ndim)
print("shape:", zeros_arr.shape)
print("size:", zeros_arr.size)

sample_arr = np.array([1, 2.5, 3])
print("dtype:", sample_arr.dtype)
```
